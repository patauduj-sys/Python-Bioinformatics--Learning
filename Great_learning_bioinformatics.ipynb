````python
import random
bases = ["A", "T", "G", "C"]
random_dna = "".join(random.choices(bases, k=30))
````

````python
from collections import Counter

# Count the frequency of each base in the sequence
base_counts = Counter(random_dna)

# Display the frequencies
for base, count in base_counts.items():
    print(f"{base}: {count}")
````

````python
# Calculate the percentage of each base
total_bases = len(random_dna)
print(f"Total bases: {total_bases}\n")

for base, count in base_counts.items():
    percentage = (count / total_bases) * 100
    print(f"{base}: {percentage:.2f}%")
````

### Understanding the `input()` Function

The `input()` function is used to take user input from the keyboard.

#### Key Features:
1. **Pauses Execution**: When Python reaches `input()`, it stops and waits for the user to type something and press **Enter**.
2. **Always Returns a String**: No matter what the user types (letters, words, or numbers), `input()` always treats it as text (a `str`).
3. **Optional Prompt**: You can put a message inside `input("...")` to tell the user what they should enter.

#### Examples:

* **Getting Text Input:**
  ```python
  user_color = input("What is your favorite color? ")
  print("Your favorite color is", user_color)
  ```

* **Getting Numeric Input (Requires Casting):**
  Because `input()` returns a string, if you want to perform math on a number the user typed, you must convert it using `int()` or `float()`:
  ```python
  age_str = input("Enter your age: ")
  age = int(age_str)  # Converts the string to an integer
  print("Next year you will be", age + 1)
  ```

### Celsius to Fahrenheit Converter

This program prompts the user for a temperature in Celsius, converts it to Fahrenheit using the formula:

$$F = (C \times \frac{9}{5}) + 32$$

````python
# Prompt the user for Celsius temperature
celsius = float(input("Enter temperature in Celsius: "))

# Convert to Fahrenheit
fahrenheit = (celsius * 9/5) + 32

# Display the converted temperature
print(f"{celsius}°C is equal to {fahrenheit}°F")
````
