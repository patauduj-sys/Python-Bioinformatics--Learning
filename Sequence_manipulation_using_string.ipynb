#counting the number of characters  e.g count the numbers of a, t,etc

```python
a_count_dna1 = dna1.count("A")
print(a_count_dna1)
```

```python
t_count_dna2 = dna2.count("t")
print(t_count_dna2)
```

#finding gc content

```python
length_dna3 = len(dna3)
g_count_dna3 = dna3.count("G")
c_count_dna3 = dna3.count("C")
gc_content = g_count_dna3 + c_count_dna3
percentage_gc_content = (gc_content / length_dna3) * 100
print(percentage_gc_content)
```

```python
length = len(dna2)
g_count_dna2 = dna2.count("g")
c_count_dna2 = dna2.count("c")
gc_content  = g_count_dna2 + c_count_dna2
percentage_gc_content = (gc_content / length) * 100
print(percentage_gc_content)
```
